# Maneul Olarak Gradient Boosting

- Bu gerçek değil aslında tüm parametrelerle oynamıyoruz sadece birini değiştiriyoruz ama buradaki asıl amaç mantık nasıl oluyor bunu kavramak.

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('Strength', axis=1)
y = df.Strength
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=36)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
# first weark learner
tree_reg1 = DecisionTreeRegressor(max_depth=3)
tree_reg1.fit(X_train, y_train)
```

```python
# residual error
y2 = y_train - tree_reg1.predict(X_train)
y2[:5]
```

```python
# second weak learner
tree_reg2 = DecisionTreeRegressor(max_depth=4)
tree_reg2.fit(X_train, y2)
```

```python
y3 = y2 - tree_reg2.predict(X_train)
y3[:5]
```

```python
#third weak learner
tree_reg3 = DecisionTreeRegressor(max_depth = 4)
tree_reg3.fit(X_train, y3)
```

```python
y4 = y3 - tree_reg3.predict(X_train)
y4[:5]
```

```python
# let's calculate the final score (we can obviously increase the number of weak learners if we want)
y_pred = sum(tree.predict(X_test) for tree in (tree_reg1, tree_reg2, tree_reg3))
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test, y_pred)
```

# Otomatik Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
gbr = GradientBoostingRegressor()
```

```python
gbr.fit(X_train, y_train)
```

```python
y_pred = gbr.predict(X_test)
r2 = r2_score(y_test, y_pred)
print(r2)
```

```python
params = {
    'loss' : ['squared_error', 'absolute_error', 'huber', 'quantile'],
    'learning_rate' : [0.01, 0.1, 1],
    'n_estimators' : [1, 10, 50, 100, 200, 500],
    'criterion' : ['friedman_mse', 'squared_error']
}
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
import warnings
warnings.filterwarnings('ignore')
rscv = RandomizedSearchCV(estimator=GradientBoostingRegressor(), param_distributions=params, verbose=3, refit=True)
```

```python
rscv.fit(X_train, y_train)
```

```python
rscv.best_params_
```

```python
gradient = GradientBoostingRegressor(n_estimators=500, loss='huber', learning_rate=0.1, criterion='friedman_mse')
gradient.fit(X_train, y_train)
```

```python
y_pred = gradient.predict(X_test)
r2 = r2_score(y_test, y_pred)
print(r2)
```
