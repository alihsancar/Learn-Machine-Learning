```python
df['age'] = df['age'].fillna(df['age'].median())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
df['adult_male'] = df['adult_male'].astype(int)
df['alone'] = df['alone'].astype(int)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('survived', axis =1)
y = df['survived']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=36)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
onehot_columns = ['sex', 'embarked', 'who', 'class']
```

```python
transformer = ColumnTransformer(
    transformers=[
        ('onehot', OneHotEncoder(drop = 'first', handle_unknown='ignore'), onehot_columns)
    ],
    remainder='passthrough'
)
```

```python
X_train = transformer.fit_transform(X_train)
X_test = transformer.transform(X_test)
```

```python
new_cols = transformer.get_feature_names_out()
```

```python
new_cols
```

```python
X_train = pd.DataFrame(data=X_train, columns=new_cols)
X_test = pd.DataFrame(data=X_test, columns=new_cols)
```

```python
X_train.head()
```

```python
X_test.head()
```

```python
X_train.info()
```

```python
import lightgbm as lgb
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
light = lgb.LGBMClassifier(verbosity=-1)
```

```python
light.fit(X_train, y_train)
```

```python
y_pred = light.predict(X_test)
```

```python
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred, y_test))
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(light.predict(X_train), y_train))
```

```python
importances = light.feature_importances_

feature_names = X_train.columns

feat_importance = pd.DataFrame(
    {
        'Feature' : feature_names,
        'Importance' : importances
    }
).sort_values(by='Importance',ascending=False)

print(feat_importance)
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
params = {
    'n_estimators' : [100, 300, 500, 1000],
    'num_leaves' : [5, 31, 63, 127, 511],
    'max_depth' : [2, 3, 4, 5, 10, 15, 30, 50, None],
    'learning_rate' : [0.01, 0.1, 1],
    'min_child_samples': [5, 10, 20],
    'subsample': [0.6, 0.8, 1.0],
    'colsample_bytree': [0.6, 0.8, 1.0]
}
```

```python
random = RandomizedSearchCV(lgb.LGBMClassifier(verbosity = -1), param_distributions=params, cv=5, refit=True, verbose=1)
```

```python
random.fit(X_train, y_train)
```

```python
y_pred = random.predict(X_test)
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred, y_test))
```
