# Data Cleaning and Encoding

```python
def ouitliers_iqr(df, threshold):
    outliers_summary = {}
    numeric_col = [col for col in df.columns if df[col].dtype != 'O']

    for col in numeric_col:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - IQR * threshold
        upper = Q3 + IQR * threshold

        outliers = df[ (df[col] < lower) | (df[col] > upper) ]
        outliers_summary[col] = {
            'outliers_shape' : outliers.shape[0],
            'outliers_percantage' : 100 * outliers.shape[0] / df.shape[0],
            'lower' : lower,
            'upper' : upper
        }

    return pd.DataFrame(outliers_summary)
```

```python
ouitliers_iqr(df, 1.5)
```

```python
def remove_outliers_from_all(df, threshold):
    df_clean = df.copy()
    numeric_col = [col for col in df_clean.columns if df_clean[col].dtype != 'O']

    for col in numeric_col:
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - IQR * threshold
        upper = Q3 + IQR * threshold

        df_clean = df_clean[ (df_clean[col] >= lower) & (df_clean[col] <= upper) ]
    return df_clean.copy()
```

```python
print("Önceki : ", df.shape)
df_clean = remove_outliers_from_all(df, 1.5)
print("Sonraki : ", df_clean.shape)
```
